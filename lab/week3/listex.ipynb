## List 연습 노트 파일입니다.

리스트의 합병 연산 (+)
리스트의 곱 연산 (*)

```python
fruits1 = ["사과", "바나나"]
fruits2 = ["복숭아", "포도"]
fruits = fruits1 + fruits2 

print(fruits)
```

리스트 얇은 복사1

```python
temps = [28, 31, 33, 35, 27, 26, 25]
values = list(temps)
temps[0] = 50
temps
values
```

리스트 얕은 복사2

```python
temps = [ [1,2], 3, 4, 5]
values = list(temps)
temps[-1] = 6
temps
values
temps[0][0] = 7
temps
values
```

리스트 깊은 복사

```python
import copy
temps = [ [1,2], 3, 4, 5]
svalues = copy.copy(temps)
temps[0][0] = 100
temps
svalues
dvalues = copy.deepcopy(temps)
temps[0][0] = 7
temps
dvalues
```
