# 객채지향 학습

- 객채지항이란?
- 클래스란?
- 클래스의 인스턴스 = 객채

```python
class Counter:
    def __init__(self, count=0):
        self.count = count
    
    def increment(self):
        self.count += 1

    def get_count(self):
        return self.count


def test_counter():
    myc = Counter()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    for _ in range(10):
        myc.increment()
    print(myc.get_count())
   
    yourc = Counter(5)
    print(yourc.get_count())
    yourc.increment()
    print(yourc.get_count())
    yourc.increment()
    print(yourc.get_count())


if __name__ == "__main__":
    test_counter()
```

# TV 클래스 추가

- 데이터: 현재 채널번호 볼륨량 on여부
- 동작: 켜기 끄기 채널변경 볼륨 변경 현재 TV 객채 내용 출력
- 클래스명: MyTelevision

```python
class MyTelevision:
    def __init__(self, channel = 1, volume = 10, is_on = False):
        self.channel = channel
        self.volume = volume
        self.is_on = is_on
        MyTelevision.serialNumber += 1
        self.serial_Number = MyTelevision.serialNumber
    
    def Turn_on(self):
        self.is_on = True

    def Turn_off(self):
        self.is_on = False

    def change_channel(self, channel):
        self.channel = channel 

    def change_volume(self, volume):
        self.volume = volume

    def display_status(self):
        print(f"channel: {self.channel}, Volume: {self.volume}, Is On: {self.is_on}, Serial Number: {self.serial_Number}")

def test_television():
    mytv = MyTelevision(channel = 11)
    mytv.Turn_on()
    mytv.display_status()
    mytv.Turn_off()
    mytv.display_status()

test_television()
```

# TV 리모콘 클래스
- 데이터는 없고 동작만 있음
- 연결된 TV를 키거나 끼라는 동작만 지원
- TV 상태 표시 추가 

```python
class TVremocon :
    def __init__(self, tv:MyTelevision):
        self.tv = tv

    def turn_on(self):
        self.tv.Turn_on()

    def turn_off(self):
        self.tv.Turn_off()

    def press_display(self):
        self.tv.display_status()

def test_remocon():
    tv1 = MyTelevision(1, 5, False)
    rc1 = TVremocon(tv1)
    rc1.turn_on()
    rc1.press_display()
    rc1.turn_off()
    rc1.press_display()


test_remocon()
```
