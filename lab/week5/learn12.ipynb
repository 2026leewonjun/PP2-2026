- 상속에 대한 이론과 실재 학슴

```python
class car:
    def __init__(self, maker, model, color, price):
        self.maker = maker
        self.model = model
        self.color = color 
        self.price = price

    def setMaker(self, maker):
        self.maker = maker

    def getMaker(self):
        return self.maker

    def getDesc(self):
        return self.__str__()

    def __str__(self):
        return f"Maker: {self.maker}, Model: {self.model}, Color: {self.color}, Price: {self.price}"
    
class ElectricCar(car):
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price)
        self.batterySize = bat_size

    def getBatterySize(self):
        return self.batterySize

    def setBatterySize(self, bat_size):
        self.batterySize = bat_size
    
    def __str__(self):
        return f"{super().__str__()}, Battery: {self.batterySize}kwh"

    def getDesc(self):
        return self.__str__()

class Truck(car):
    def __init__(self, maker, model, color, price, playLoad):
        super().__init__(maker, model, color, price)
        self.playLoad = playLoad

    def __str__(self):
        return f"{super().__str__()}, Payload: {self.playLoad}kg"
    
car1 = car("kia", "Ray", "Blue", 2000)
print(car1.getDesc())
print(car1.model)

ecar1 = ElectricCar("Tesla", "Model S", "Red", 8000, 81)
print(ecar1)
ecar1.setBatterySize(92)
print(ecar1.getDesc())
truc1 = Truck("Ford", "F-150", "Black", 4000, 1000)
print(truc1)
print(truc1.getMaker())

print(type(truc1))
print(type(ecar1))
print(type(car1))

print(isinstance(truc1, Truck))
print(isinstance(ecar1, ElectricCar))
print(isinstance(car1, car))

print(isinstance(truc1, car))
print(isinstance(ecar1, car))
```

# 다중 상속: 2개 이상의 부모클래스로부터 상속받을 떄

```python
class Base1:
    def __init__(self, v1 = 10):
        self.attr1 = v1

    def display1(self):
        print(f"Base1: {self.attr1}")

    def display(self):
        self.display1()

class Base2:
    def __init__(self, v2 = 20):
        self.attr2 = v2

    def display2(self):
        print(f"Base2: {self.attr2}")

    def display(self):
        self.display2()

class C(Base1, Base2):
    def __init__(self, v1, v2, v3):
        Base1.__init__(self, v1)
        Base2.__init__(self, v2)
        self.attr3 = v3

    def print(self):
        print(f"{self.attr1}, {self.attr2}, {self.attr3}")
        self.display1()
        self.display2()
        self.display()
        Base1.display(self)
        Base2.display(self)

def test():
    c = C(v1=1, v2=2, v3=3)
    c.print()

    c2 = C(4,5,6)
    c2.print()

if __name__ == "__main__":
    test()
```

# Overriding: 부모의 함수(메소드) 재정의

```python
import math

class Shape:
    pass

class Circle(Shape):
    def __init__(self, name="도형"):
        self.name = name

    def draw(self):
        print (f"(self.name)을(를) 그립니다.")

    def get_area(self):
        print(f"Shape: {self.name}의 면적을 계산합니다.")

class Circle(Shape):
    def __init__(self, radius=1, name="원"):       
        super().__init__(name)
        self.radius = radius

    def draw(self):
        print(f"CIrcle: {self.name}의 면적을 계산합니다.")
        return math.pi * self.radius ** 2

def test():
    c = Circle()
    c.draw()
    print(c.get_area())

if __name__ == "__main__":
    test()
```

- LAB 직원과 매니저


```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def getSalary(self):
        return self.salary

class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = bonus

    def getSalary(self):
        salary = super().getSalary()
        return salary + self.bonus

    def __repr__(self):
        return "이름: " + self.name + "; 월급: " + str(self.salary) + \
               "; 보너스: " + str(self.bonus)

kim = Manager("김철수", 2000000, 1000000)
print(kim)
```

- LAB Card와 Deck

```python
class Card:
    suitNames = ['클럽', '다이아몬드', '하트', '스페이드']
    rankNames = [None, '에이스', '2', '3', '4', '5', '6', '7',
                 '8', '9', '10', '잭', '퀸', '킹']

    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank

    def __str__(self):
        return Card.suitNames[self.suit] + " " + \
               Card.rankNames[self.rank]

class Deck:
    def __init__(self):
        self.cards = []
        for suit in range(4):
            for rank in range(1, 14):
                card = Card(suit, rank)
                self.cards.append(card)

    def __str__(self):
        lst = [str(card) for card in self.cards]
        return str(lst)

deck = Deck()
print(deck)
```
