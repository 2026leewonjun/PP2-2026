- 매출 파일 처리


```python
infilename = input("입력 파일 이름: ");
outfilename = input("출력 파일 이름: ");

infile = open(infilename, "r")
outfile = open(outfilename, "w")
sum = 0
count = 0

line = infile.readline()
while line != "":
    s = sum(line)
    sum += s
    count += 1
    line = infile.readline()
outfile.write("총매출: " + str(sum) + "\n")
outfile.write("평균 일매출: " + str(sum / count) + "\n")

infile.close()
outfile.close()
```

- 다양한 텍스트 파일 입출력 방법


```python
counter = [0] *26
infile = open("mobydick.txt", "r")
ch = infile.read(1)
while ch != "":
    ch = ch.upper()
    if ch >= 'A' and ch <= 'Z':
        i = ord(ch) - ord('A')
        counter[i] += 1
    ch = infile.read(1)
print(counter)
```

- 디렉토리 작업

```python
import os

cwd = os.getcwd()
files = os.listdir(cwd)
for name in files :
    if os.path.isfile(name) :
        if name.endswith(".jdg") :
            print(name)
```
