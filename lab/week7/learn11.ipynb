- LAB 키를 이용한 정렬 예제

```python
class Person(object):
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __repr__(self):
        return "<이름: %s,  나이: %s>" % (self.name, self.age)


def keyAge(person):
    return person.age

people = [Person("홍길동", 20), Person("김철수", 35), Person("최자영", 38)]
print(sorted(people, key=keyAge))
```

- 람다식으로 온도 변환하기

```python
f_temp = [0, 10, 20, 30, 40, 50]
c_temp = list(map(lambda x: (5.0/9.0) * (x - 32), f_temp))
print(list(c_temp))
```

- 람다식으로 데이터 처리하기

```python
order = [   [ "1", "재킷", 5, 120000],
            ["2", "셔츠", 6, 24000],
            ["3", "바지", 3, 50000],
            ["4", "코트", 6, 300000] ]

result = list(map(lambda x: (x[0], x[2] * x[3]), order))
print(result)
```

- 피보나치 이터레이터

```python
class Fiblterator:
    def __init__(self, a=1, b=0, maxValue=50):
        self.a = a
        self.b = b
        self.maxValue = maxValue

    def __iter__(self):
        return self

    def __next__(self):
        n = self.a + self.b
        if n > self.maxValue:
            raise StopIteration
        self.a = self.b
        self.b = n
        return n
    for i in Fiblterator():
        print(i, end=' ')
```
